# Comprehension (내포, 함축)

- 한 줄 짜리 반복문으로 새로운 컬렉션(list, dict..)을 생성하는 문법

- 반복문으로 값을 하나씩 추가하는 패턴을 더 간략히 만들고, 읽기 좋게 표현할 수 있다

- list comprehension
```python
# list 생성
[표현식 for 변수 in 반복가능한객체]

# 조건 포함
[표현식 for 변수 in 반복가능한객체 if 조건식]
```

- dict comprehension
```python
# dict 생성
{key표현식: value표현식 for 변수 in 반복가능한객체}

# 조건 포함
{key표현식: value표현식 for 변수 in 반복가능한객체 if 조건식}
```

## 1. list comprehension

````python
# 일반 for문 + list
numbers: list[int] = []
for n in range(1, 6):
    numbers.append(n)

print(numbers)
````

````python
# list 생성 함축 구문 사용

# 1. [] 내에 for문 그대로 작성
# 2. for문에서 요소 하나를 저장한 변수(n)을 꺼내
#    'for' 왼쪽에 배치 == []에 추가(append())
numbers: list[int] = [  (n + 1) for n in range(1, 6)  ]

print(numbers)
````

````python
# 조건식이 포함된 list comprehension

# 1~30 사이 정수 중 3의 배수만 list에 저장
# 1) 일반 표기법
numbers: list[int] = []

for n in range(1,31,1):
    if n % 3 == 0:
        numbers.append(n)

print(numbers)
````

````python
# 2) 조건식이 포함된 comprehension

# 1. [] 내에 for문 그대로 작성
# 2. for문에서 요소 하나를 저장한 변수(n)을 꺼내 
#    for문 오른쪽 'if문에' 전달
# 3. if문 조건식 결과가 True인 경우에만 
#    'for' 왼쪽에 배치 == []에 추가하다(append())
numbers: list[int] = [ n for n in range(1,31,1) if n % 3 == 0 ]

print(numbers)
````
